```python
import lightkurve as lk
import matplotlib.pyplot as plt
```

```python
search_result = lk.search_lightcurve(
    "Kepler-10",
    mission="Kepler"
)

lcs = search_result.download_all()

lc = lcs.stitch()
```

```python
lc = lc.remove_nans()
```

```python
lc.plot()
plt.show()
```

```python
flat_lc = lc.flatten(window_length=401)
```

```python
flat_lc.plot()
plt.show()
```

```python
fig, ax = plt.subplots(2, 1, figsize=(12, 8))

lc.plot(ax=ax[0])
ax[0].set_title("Raw Light Curve")

flat_lc.plot(ax=ax[1])
ax[1].set_title("Flattened Light Curve")

plt.tight_layout()
plt.show()
```
